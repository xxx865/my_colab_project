```python
from pathlib import Path
import pymupdf
pdf_texts = {}
for path in Path("/content").glob("*.pdf"):
  with pymupdf.open(path) as doc:
    pdf_texts[path.name] = "\n".join(
        page.get_text() for page in doc
    )
print("已读取：",list(pdf_texts))
```

```python
from langchain_community.document_loaders import PyPDFLoader
loaders = [
    PyPDFLoader("/content/MachineLearning-Lecture01.pdf"),
    PyPDFLoader("/content/MachineLearning-Lecture01.pdf"),
    PyPDFLoader("/content/MachineLearning-Lecture02.pdf"),
    PyPDFLoader("/content/MachineLearning-Lecture03.pdf"),
    PyPDFLoader("/content/MachineLearning-Lecture04.pdf")
]
docs = []
for loader in loaders:
  docs.extend(loader.load())
```
