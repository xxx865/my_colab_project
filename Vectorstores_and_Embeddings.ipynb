```python
from pathlib import Path
import pymupdf
pdf_texts = {}
for path in Path("/content").glob("*.pdf"):
  with pymupdf.open(path) as doc:
    pdf_texts[path.name] = "\n".join(
        page.get_text() for page in doc
    )
print("已读取：",list(pdf_texts))
```

```python
%pip -q install -U langchain-huggingface sentence-transformers
```

```python
from langchain_huggingface import HuggingFaceEmbeddings

embedding = HuggingFaceEmbeddings(
    model_name="sentence-transformers/paraphrase-multilingual-MiniLM-L12-v2",
    model_kwargs={"device": "cpu"},
    encode_kwargs={"normalize_embeddings": True},
)
```

```python
vector = embedding.embed_query("这篇论文研究了什么")
print("向量维度：",len(vector))
```

```python
sentences1="i like dogs"
sentences2="i like canines"
sentences3="the weather is ugly outside"
```

```python
embeddings1 = embedding.embed_query(sentences1)
embeddings2 = embedding.embed_query(sentences2)
embeddings3 = embedding.embed_query(sentences3)
```

```python
sentence1_chinese = "我喜欢狗"
sentence2_chinese = "我喜欢犬科动物"
sentence3_chinese = "外面的天气很糟糕"
```
