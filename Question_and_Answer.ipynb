```python
from langchain_deepseek import ChatDeepSeek
from langchain_core.prompts import ChatPromptTemplate

llm = ChatDeepSeek(
    model="deepseek-chat",
    temperature=0,
)

prompt = ChatPromptTemplate.from_template(
    "请介绍一下：{topic}"
)

chain = prompt | llm

result = chain.invoke({"topic": "LangChain"})
print(result.content)
```

```python
from sentence_transformers import SentenceTransformer
import numpy as np

model = SentenceTransformer('BAAI/bge-small-zh-v1.5')

texts = ["如何学习Python", "Python入门方法", "今天天气真好"]
vecs = model.encode(texts)

# 计算余弦相似度
def cos(a, b):
    return np.dot(a, b) / (np.linalg.norm(a) * np.linalg.norm(b))

print(cos(vecs[0], vecs[1]))  # 约 0.85，很相似
print(cos(vecs[0], vecs[2]))  # 约 0.20，不相似
```

```python
import chromadb

client = chromadb.Client()
collection = client.create_collection("my_knowledge_V2")

# 存入（Chroma 可自动调 embedding 函数，也可自己传向量）
collection.add(
    documents=["Python是一种编程语言", "Java也是编程语言", "猫是一种动物"],
    ids=["1", "2", "3"]
)

# 查询
results = collection.query(query_texts=["哪些是编程语言"], n_results=2)
print(results["documents"])
```

```python
import chromadb
from chromadb.utils import embedding_functions

# 换更大的模型
ef = embedding_functions.SentenceTransformerEmbeddingFunction(
    model_name="BAAI/bge-base-zh-v1.5"
)

client = chromadb.Client()
client.delete_collection("my_knowledge")   # 换模型后重建
collection = client.get_or_create_collection(
    name="my_knowledge",
    embedding_function=ef
)

collection.add(
    documents=["Python是一种编程语言", "Java也是编程语言", "猫是一种动物"],
    ids=["1", "2", "3"]
)

results = collection.query(
    query_texts=["哪些是编程语言"],
    n_results=3,
    include=["documents", "distances"]
)

print("全部结果（按相似度排序）：")
for doc, dist in zip(results["documents"][0], results["distances"][0]):
    print(f"  distance={dist:.4f}  {doc}")

# 加阈值过滤
print("\n过滤后：")
for doc, dist in zip(results["documents"][0], results["distances"][0]):
    if dist < 0.6:
        print(f"  {doc}")
```
