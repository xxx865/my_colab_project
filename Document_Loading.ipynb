```python
from google.colab import files
from langchain_community.document_loaders.pdf import PyPDFLoader
uploaded = files.upload()
pdf_path = next(iter(uploaded))
loader = PyPDFLoader(pdf_path)
pages = loader.load()
print(f"读取了{len(pages)}页")
```
